# Nodes in LangGraph
Companion notebook for the **Nodes in LangGraph** section of Chapter 4.

Nodes are the processing units: they read from state, do the work (I/O, LLM calls, compute), and return an update to the state.

```python
from pydantic import BaseModel, Field


class ResearchState(BaseModel):
    query: str = ""
    candidate_urls: list[str] = Field(default_factory=list)
    page_texts: list[str] = Field(default_factory=list)
    draft_summary: str = ""
    final_report: str = ""
    errors: list[str] = Field(default_factory=list)
```

The book calls out an LLM call with `llm_summarize(text)`. Below is a tiny stand-in for it -- notebook 04 shows a node that performs a real LLM call.

```python
def llm_summarize(text: str) -> str:
    """Placeholder for an LLM call (see notebook 04 for a real one)."""
    return f"Summary of: {text}"


def summarize(state: ResearchState) -> ResearchState:
    text = state.query
    state.draft_summary = llm_summarize(text)  # an LLM call
    return state
```

```python
state = ResearchState(query="Latest advances in battery technology")
updated_state = summarize(state)
print("draft_summary:", updated_state.draft_summary)
```
