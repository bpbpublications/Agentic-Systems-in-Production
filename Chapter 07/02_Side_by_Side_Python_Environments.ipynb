# Side-by-Side Python Environments
Companion notebook for the **Side-by-Side Python Environments** section of Chapter 7.

This notebook actually creates an isolated virtual environment next to the scaffold, installs `agent-app` into it, and proves the install is isolated -- the same steps the book shows as shell commands, run for real via `!` shell cells.

```python
!python -m venv .venv_demo
```

Install the package into that isolated environment. On Windows this would normally be `.venv_demo\Scripts\pip.exe`; the path below picks whichever layout exists so this cell works the same on Windows, Linux, and Mac.

```python
import sys
from pathlib import Path

venv_python = Path(".venv_demo") / ("Scripts/python.exe" if sys.platform == "win32" else "bin/python")
print(venv_python)
```

```python
!"{venv_python}" -m pip install --quiet -e ./agent-app
```

Prove the install is real and isolated -- import the package using *only* the new environment's interpreter, not this notebook's own kernel:

```python
!"{venv_python}" -c "import support_agent; print('installed version:', support_agent.__version__)"
```

`pyproject.toml` keeps runtime dependencies small and puts developer/test tools in **extras**, installed with `.[dev]` or `.[test]`:

Clean up the throwaway environment (in a real project you'd keep `.venv/` -- and add it to `.gitignore` -- rather than deleting it after every install):
